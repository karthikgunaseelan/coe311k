```python
import numpy as np
import matplotlib.pyplot as plt

# Define the parameters

k = 1.0        # Constant in the equation (e.g., spring constant)
h = 0.01       # Step size
x_start = 0    # Start of the interval
x_end = 10     # End of the interval
y_0 = 1.0      # Initial condition for y
v_0 = 0.0      # Initial condition for dy/dx

# Create arrays to store x, y, and v values

x_values = np.arange(x_start, x_end + h, h)
y_values = np.zeros(len(x_values))
v_values = np.zeros(len(x_values))

# Set initial conditions

y_values[0] = y_0
v_values[0] = v_0

# Euler's Method loop

for i in range(1, len(x_values)):
    # Current values
    x_n = x_values[i - 1]
    y_n = y_values[i - 1]
    v_n = v_values[i - 1]    

    # Update equations
    y_values[i] = y_n + h * v_n
    v_values[i] = v_n + h * (-k * y_n)

# Plot the results

plt.figure(figsize=(10, 5))
plt.plot(x_values, y_values, label='Position y(x)')
plt.plot(x_values, v_values, label='Velocity v(x)', linestyle='--')
plt.xlabel('x')
plt.ylabel('y and v')
plt.title('Solution of d²y/dx² = -ky using Euler\'s Method')
plt.legend()
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the parameters

k = 1.0        # Constant in the equation (e.g., spring constant)
h = 0.01       # Step size
x_start = 0    # Start of the interval
x_end = 10     # End of the interval
y_0 = 1.0      # Initial condition for y
v_0 = 0.0      # Initial condition for dy/dx
h_values = np.array([.01, .02, .03, .05, .1, .2, .3, .5])

# Create loop to store x, y, and v values

def defineParameters(h):
   
    # Create arrays to store x, y, and v values
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))
    
    # Set initial conditions
    y_values[0] = y_0
    v_values[0] = v_0
    
    return x_values, y_values, v_values

fig, axs = plt.subplots(2, 1, figsize=(10, 6)) #Looked up

for h in h_values:
    
    x_values, y_values, v_values = defineParameters(h)
    
    for i in range(1, len(x_values)):
        
        # Current values
        x_n = x_values[i - 1]
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]    

        # Update equations
        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)
    
    axs[0].plot(x_values, y_values, label=f'h={h}')
    axs[1].plot(x_values, v_values, label=f'h={h}')
    
axs[0].set_title('Position')
axs[1].set_title('Velocity')
axs[0].set_xlabel('x')
axs[1].set_xlabel('x')
axs[0].set_ylabel('y')
axs[1].set_ylabel('v')

for ax in axs:
    ax.legend()
```

With larger step sizes h, Euler's Method becomes more unstable, where it becomes very clear at a step size of .2.
